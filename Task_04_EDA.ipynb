## Exploratory Data Analysis (EDA)

This notebook performs descriptive statistics, visualizations, correlation analysis, 3 business hypothesis tests, and Top 5 findings.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display
from scipy.stats import pearsonr, ttest_ind
```

## 1. Load the cleaned dataset


```python
df = pd.read_csv("clean_dataset.csv")
print("Dataset shape:", df.shape)
display(df.head())
```

## 2. Descriptive statistics


```python
display(df[["Sales","Discount","Profit","Profit_Margin","Customer_Age"]].describe())
```

## 3. Mean, median, standard deviation and quartiles


```python
display(df[["Sales","Discount","Profit","Profit_Margin","Customer_Age"]].agg(["mean","median","std"]))
print("Quartiles:")
display(df[["Sales","Discount","Profit","Profit_Margin","Customer_Age"]].quantile([.25,.50,.75]))
```

## 4. Histograms


```python
df["Sales"].hist(bins=30)
plt.title("Distribution of Sales"); plt.xlabel("Sales"); plt.ylabel("Frequency"); plt.show()
```

## 4B. Profit histogram


```python
df["Profit"].hist(bins=30)
plt.title("Distribution of Profit"); plt.xlabel("Profit"); plt.ylabel("Frequency"); plt.show()
```

## 4C. Discount histogram


```python
df["Discount"].hist(bins=20)
plt.title("Distribution of Discount"); plt.xlabel("Discount"); plt.ylabel("Frequency"); plt.show()
```

## 5. Box plots - Sales


```python
plt.boxplot(df["Sales"]); plt.title("Sales Box Plot"); plt.ylabel("Sales"); plt.show()
```

## 5B. Box plot - Profit


```python
plt.boxplot(df["Profit"]); plt.title("Profit Box Plot"); plt.ylabel("Profit"); plt.show()
```

## 6. Correlation heatmap


```python
corr=df[["Sales","Discount","Profit","Profit_Margin","Customer_Age"]].corr()
plt.figure(figsize=(8,6)); plt.imshow(corr,aspect="auto"); plt.colorbar(label="Correlation")
plt.xticks(range(len(corr.columns)),corr.columns,rotation=45,ha="right"); plt.yticks(range(len(corr.columns)),corr.columns)
plt.title("Correlation Heatmap"); plt.tight_layout(); plt.show()
```

## 7. Hypothesis 1
**H0:** Discount has no significant relationship with Profit Margin.

**H1:** Higher discount is associated with lower Profit Margin.


```python
r,p=pearsonr(df["Discount"],df["Profit_Margin"])
print(f"Correlation = {r:.4f}, p-value = {p:.6f}")
print("Conclusion:", "Reject H0: significant relationship." if p<.05 else "Fail to reject H0.")
```

## 8. Hypothesis 2
**H0:** Average Sales are equal for retained and non-retained customers.

**H1:** Average Sales differ.


```python
a=df.loc[df["Customer_Retained"]=="Yes","Sales"]
b=df.loc[df["Customer_Retained"]=="No","Sales"]
t,p=ttest_ind(a,b,equal_var=False)
print("Retained average sales:",round(a.mean(),2)); print("Non-retained average sales:",round(b.mean(),2))
print(f"t-statistic={t:.4f}, p-value={p:.6f}")
print("Conclusion:", "Reject H0: significant difference." if p<.05 else "Fail to reject H0.")
```

## 9. Hypothesis 3
Compare average Profit across product categories.


```python
cat_profit=df.groupby("Category")["Profit"].agg(["mean","median","std","count"]).sort_values("mean",ascending=False)
display(cat_profit)
print("Highest average-profit category:",cat_profit.index[0])
print("Lowest average-profit category:",cat_profit.index[-1])
```

## 10. Additional business analysis


```python
print("Sales by category:"); display(df.groupby("Category")["Sales"].agg(["sum","mean","count"]).sort_values("sum",ascending=False))
print("Profit by region:"); display(df.groupby("Region")["Profit"].agg(["sum","mean","count"]).sort_values("sum",ascending=False))
```

## 11. Top 5 Business Findings


```python
findings=[]
findings.append(f"{cat_profit.index[0]} has the highest average profit.")
findings.append(f"{cat_profit.index[-1]} has the lowest average profit.")
findings.append(f"Discount vs profit-margin correlation is {r:.2f}.")
findings.append(f"Retained customers average sales: {a.mean():.2f}; non-retained: {b.mean():.2f}.")
best_region=df.groupby("Region")["Profit"].sum().idxmax()
findings.append(f"{best_region} has the highest total regional profit.")
for i,x in enumerate(findings,1): print(f"{i}. {x}")
```
